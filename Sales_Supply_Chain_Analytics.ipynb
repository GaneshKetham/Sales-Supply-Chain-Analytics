```python
year_analysis = df.groupby('Year').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).reset_index()

year_analysis
```

```python
monthly_analysis = df.groupby('Month').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).reset_index()

monthly_analysis
```

```python
category_analysis = df.groupby('Category Name').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).sort_values('Sales', ascending=False).reset_index()

category_analysis
```

```python
department_analysis = df.groupby('Department Name').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).sort_values('Sales', ascending=False).reset_index()

department_analysis
```

```python
segment_analysis = df.groupby('Customer Segment').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).sort_values('Sales', ascending=False).reset_index()

segment_analysis
```

```python
market_analysis = df.groupby('Market').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).sort_values('Sales', ascending=False).reset_index()

market_analysis
```

```python
region_analysis = df.groupby('Order Region').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).sort_values('Sales', ascending=False).reset_index()

region_analysis
```

```python
delivery_analysis = df.groupby('Delivery Status').agg(
    Orders=('Order Id', 'nunique'),
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum')
).reset_index()

delivery_analysis
```

```python
shipping_analysis = df.groupby('Shipping Mode').agg(
    Orders=('Order Id', 'nunique'),
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Avg_Shipping_Days=('Days for shipping (real)', 'mean'),
    Late_Rate=('Late_delivery_risk', 'mean')
).reset_index()

shipping_analysis['Late_Rate'] = shipping_analysis['Late_Rate'] * 100

shipping_analysis
```

```python
delivery_gap_shipping = df.groupby('Shipping Mode').agg(
    Avg_Delivery_Gap=('Delivery Gap', 'mean'),
    Orders=('Order Id', 'nunique'),
    Late_Rate=('Late_delivery_risk', 'mean')
).reset_index()

delivery_gap_shipping['Late_Rate'] = delivery_gap_shipping['Late_Rate'] * 100

delivery_gap_shipping
```

```python
region_delivery = df.groupby('Order Region').agg(
    Orders=('Order Id', 'nunique'),
    Avg_Delivery_Gap=('Delivery Gap', 'mean'),
    Late_Rate=('Late_delivery_risk', 'mean'),
    Avg_Shipping_Days=('Days for shipping (real)', 'mean')
).reset_index()

region_delivery['Late_Rate'] = region_delivery['Late_Rate'] * 100

region_delivery = region_delivery.sort_values(
    'Late_Rate', ascending=False
)

region_delivery
```

```python
profit_category = df.groupby('Category Name').agg(
    Sales=('Sales', 'sum'),
    Profit=('Order Profit Per Order', 'sum'),
    Orders=('Order Id', 'nunique')
).reset_index()

profit_category['Profit_Margin'] = (
    profit_category['Profit'] / profit_category['Sales']
) * 100

profit_category = profit_category.sort_values(
    'Profit_Margin', ascending=False
)

profit_category
```
